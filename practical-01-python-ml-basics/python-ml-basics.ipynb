```python
# Hemant Marotkar ,UID:- 23015002 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

#Load Dataset
df = pd.read_csv("basic_cyber_dataset.csv")

#Display first 5 rows
print("First 5 rows:")
print(df.head())

#Dataset shape
print("\nDatasetsize:")
print(df.shape)

#Column names
print("\nColumn names:")
print(df.columns)

#Dataset information
print("\nDataset information:")
print(df.info())

#Staistical summary
print("\nStatistical summary:")
print(df.describe())

#Missing values
print("\nMissing values:")
print(df.isnull().sum())

#Class distribution
print("\nClass distribution:")
print(df["label"].value_counts())

#Visualization
df["label"].value_counts().plot(kind="bar")
plt.title("Normal vs Attack Traffic")
plt.xlabel("Traffic Type")
plt.ylabel("Number of Records")
plt.show()

#NumPy analysis
duration_array = np.array(df["duration_ms"])
print("\nDuration Values:",duration_array)
print("Mean Duration:",np.mean(duration_array))
print("Maximum Duration:",np.max(duration_array))
print("Minimum Duration:",np.min(duration_array))

#Select features and target
X =df.drop(["connection_id","label"],axis=1)
y =df["label"]

#Split data
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

#Train model
model=DecisionTreeClassifier()
model.fit(X_train,y_train)

#Predict
y_pred=model.predict(X_test)

print("\nPredicted Values:")
print(y_pred)

print("\nActual Values:")
print(y_test.values)

#Accuracy
accuracy=accuracy_score(y_test,y_pred)
print("\nAccuracy:",accuracy)
```
