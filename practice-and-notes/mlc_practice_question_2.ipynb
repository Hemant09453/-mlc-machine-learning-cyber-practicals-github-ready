```python
import matplotlib.pyplot as plt
import numpy as np
x_values=np.array([1,2,3,4,5])
y_values=np.array([2,3,5,6,6,])
plt.plot(x_values,y_values)
plt.title("Plotting x and y values")
plt.xlabel("x values")
plt.ylabel("y values")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
xvalues=np.linspace(1,5,100)
sinvalues=np.sin(x_values)
cosvalues=np.cos(x_values)
(fig,ax)=plt.subplots(2,1)
ax[0].plot(xvalues,sinvalues,'r',label='Sin Curve')
ax[1].plot(xvalues,cosvalues,'g',label='Cos Curve')
ax[0].legend()
ax[1].legend()
ax[0].grid()
ax[1].grid(linestyle='dotted')
fig.show()
```
